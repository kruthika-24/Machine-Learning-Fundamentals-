```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df_1= pd.read_csv("House.csv")
df_2= pd.read_csv("House_price.csv")
```

```python
df_1
```

```python
df_2
```

```python
x= df_1["House Size"].to_frame()
y= df_1["House Price"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
Y_pred = model.predict(x_test)
Y_pred
```

```python
r2_score = r2_score(y_test, Y_pred)
r2_score
#print evaluation metrics
print (f"R-squared score: ",r2_score)
```

```python
# plot
plt.scatter(x, y, color='blue', label='Data points') 
plt .plot(x, model.predict(x), color='red', label='Regression Line')
plt.title("House Price vs House Size")
plt.ylabel("House Prize")
plt.xlabel('House Size')
```

```python
user_House_size = int(input("Enter House Size to predict salary: "))
user_df = pd.DataFrame(
    [[user_House_size]],
    columns=['House Size'])
predicted_salary = model.predict(user_df)
print(f"The predicted House price for House Size {user_House_size} is: {predicted_salary[0]}")
```
